```python
from PIL import Image

m = Image.open("ARMY.jpeg")
x = m.convert("L")
x.save("ARMY_image.jpeg")

y = Image.open("ARMY_image.jpeg")
plt.imshow(y)
plt.axis("off")
plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt
m=Image.open("ARMY.jpeg")
x=m.transpose(Image.FLIP_LEFT_RIGHT)
y=m.transpose(Image.FLIP_TOP_BOTTOM)
plt.subplot(311)
plt.imshow(m)

plt.subplot(312)
plt.imshow(x)

plt.subplot(313)
plt.imshow(y)

plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt
m=Image.open("ARMY.jpeg")
x=m.transpose(Image.FLIP_LEFT_RIGHT)
y=m.transpose(Image.FLIP_TOP_BOTTOM)
plt.subplot(311)
plt.imshow(m)
plt.axis("off")

plt.subplot(312)
plt.imshow(x)
plt.axis("off")

plt.subplot(313)
plt.imshow(y)
plt.axis("off")

plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt
m=Image.open("ARMY.jpeg")

plt.subplot(311)
plt.imshow(m.rotate(45))

plt.subplot(312)
plt.imshow(m.rotate(60))

plt.subplot(313)
plt.imshow(m.rotate(90))
plt.show()


```
