**Bag of words**

```python
from sklearn.feature_extraction.text import CountVectorizer
df=["I Am Ramya",
"I am good girl",
"My school name is VVM"]
model=CountVectorizer(stop_words="english")
x=model.fit_transform(df)
print(model.get_feature_names_out())

```

**TF/IDF**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
df=["I Am Ramya",
"I am good girl",
"My school name is VVM"]
model= TfidfVectorizer(stop_words="english")
x=model.fit_transform(df)
print(model.get_feature_names_out())
print(x.toarray())

```

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
df=["I Am Ramya",
"I am good girl",
"My school name is VVM"]
model=CountVectorizer(stop_words="english")
x=model.fit_transform(df)
m=pd.DataFrame(x.toarray(),columns=model.get_feature_names_out())
print(m)
```

```python
from gensim .models import Word2Vec
df=[["i","love","AI"],["AI","is","powerful"],["Machine","learning","is","fun"]]

model=Word2Vec(df,vector_size=50,window=3,min_count=1)
print(model.wv["AI"])
print(model.wv.most_similar("AI"))
```
