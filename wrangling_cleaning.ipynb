```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
from nltk.tokenize import sent_tokenize
data="""Yesterday, I went to the park with my friends, and we had a great time. The weather was warm, the sky was clear, and the birds were singing beautifully. After playing games and talking for a while, we sat under a tree and shared snacks, laughing and enjoying the moment."""
m=sent_tokenize(data)
print(m)
```

```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
from nltk.tokenize import word_tokenize
data="""Yesterday, I went to the park with my friends, and we had a great time. The weather was warm, the sky was clear, and the birds were singing beautifully. After playing games and talking for a while, we sat under a tree and shared snacks, laughing and enjoying the moment."""
m=word_tokenize(data)
print(m)
```

```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
from nltk.tokenize import regexp_tokenize
data="""Yesterday, I went to the park with my friends, and we had a great time. The weather was warm, the sky was clear, and the birds were singing beautifully. After playing games and talking for a while, we sat under a tree and shared snacks, laughing and enjoying the moment."""
m=regexp_tokenize(data,pattern="\w+")
print(m)
```
