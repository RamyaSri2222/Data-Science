```python
from sklearn.metrics import classification_report,confusion_matrix,f1_score
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression # Changed from LinearRegression
import pandas as pd

iris=load_iris()

x=iris.data
y=iris.target
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)

model=LogisticRegression(max_iter=200) # Use LogisticRegression for classification, increased max_iter for convergence
model.fit(x_train,y_train)
y_pred=model.predict(x_test) # Renamed to y_pred for consistency

print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred)) # Fixed typo in function name
print(f1_score(y_test,y_pred, average='weighted')) # Added average='weighted' for multiclass f1_score
```

```python
import pandas as pd
import numpy as np # Added for np.mean and np.std
from numpy import loadtxt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import StratifiedKFold # Corrected: StrafifiedKFold to StratifiedKFold

df=loadtxt("pima-indians-diabetes.csv",delimiter=",")
x=df[:,0:8]
y=df[:,8]

# Instantiate StratifiedKFold as a splitter object
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=1)
cvscores = [] # Renamed cvs to cvscores for clarity

# Loop through each fold
for train_index, test_index in skf.split(x, y): # Corrected: KFold.splt to skf.split and used correct indexing variables

  # Define the model inside the loop so it's re-initialized for each fold
  model = Sequential()
  # Added a hidden layer for better model capacity
  model.add(Dense(16, input_shape=(8,), activation="relu"))
  # Output layer for binary classification, should have 1 neuron and sigmoid activation
  model.add(Dense(1, activation="sigmoid"))

  model.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])

  # Fit the model on the training data for the current fold
  # Corrected: x["train"] to x[train_index] and y["train"] to y[train_index]
  # Removed validation_split as it's typically not used with explicit k-fold cross-validation
  model.fit(x[train_index], y[train_index], epochs=150, batch_size=10, verbose=0)

  # Evaluate the model on the test data for the current fold
  scores = model.evaluate(x[test_index], y[test_index], verbose=0)
  cvscores.append(scores[1] * 100) # Append accuracy, converted to percentage

# Print the average and standard deviation of the accuracy across all folds
print(f"Accuracy: {np.mean(cvscores):.2f}% (+/- {np.std(cvscores):.2f}%) errors")
```
