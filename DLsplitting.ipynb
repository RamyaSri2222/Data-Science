```python
from numpy import loadtxt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import plot_model
df=loadtxt("pima-indians-diabetes.csv",delimiter=",")
x=df[:,0:8]
y=df[:,8]
model=Sequential()

m=Dense(12,input_shape=(8,),activation="relu")
m1=Dense(8,activation="relu")
m2=Dense(1,activation="sigmoid")
model.add(m)
model.add(m1)
model.add(m2)
model.compile(loss="binary_crossentropy",optimizer="adam",metrics=["accuracy"])
model.fit(x,y,epochs=150,verbose=10,validation_split=0.2)
```

```python
from numpy import loadtxt
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import plot_model
from sklearn.model_selection import StratifiedKFold
df=loadtxt("pima-indians-diabetes.csv",delimiter=",")
x=df[:,0:8]
y=df[:,8]
kfold=StratifiedKFold(n_splits=10,shuffle=True,random_state=7)
cvscores=[]
for tarin,test in kfold.split(x,y):
   mmodel=Sequential()

   m=Dense(12,input_shape=(8,),activation="relu")
   m1=Dense(8,activation="relu")
   m2=Dense(1,activation="sigmoid")
   model.add(m)
   model.add(m1)
   model.add(m2)
   model.compile(loss="binary_crossentropy",optimizer="adam",metrics=["accuracy"])
   model.fit(x,y,epochs=150,verbose=10)
   scores=model.evaluate(x,y)
   print("%s:%.2f%%"%(model.metrics_names[1],scores[1]*100))
   cvscores.append(scores[1])
print("%.2f%%(+/-%.2f%%)"%(np.mean(cvscores),np.std(cvscores)))
```
