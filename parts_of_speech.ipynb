# **wordnet**

```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("wordnet")# word net is a module
from nltk.corpus import wordnet
m=wordnet.synsets("hello")[0]
print("syn tag:",m.pos())
print("syn tag:",m.name())
print("syn tag:",m.definition())
print("syn tag:",m.examples())

m=wordnet.synsets("beauty")[0]
print("syn tag:",m.pos())
print("syn tag:",m.name())
print("syn tag:",m.definition())
print("syn tag:",m.examples())
```

## **colloaction** **(bigrams & trigrams)**

```python
import nltk
nltk.download("punkt")
from nltk.tokenize import word_tokenize
text="i am ramya sri i am from damragidda"
l=word_tokenize(text)
print(list(nltk.bigrams(l)))
```

```python
import nltk
nltk.download("punkt")
from nltk.tokenize import word_tokenize
text="i am ramya sri i am from damragidda"
l=word_tokenize(text)
print(list(nltk.trigrams(l)))
```

```python
import nltk
nltk.download("punkt")
from nltk.tokenize import word_tokenize
text="hello ,hi good morning to all how are you ,i think all are fine and good "
l=word_tokenize(text)
print(list(nltk.bigrams(l)))

print(list(nltk.trigrams(l)))
```

```python
import nltk
nltk.download("punkt")
from nltk.tokenize import word_tokenize
from nltk import trigrams,bigrams
text="hello ,hi good morning to all how are you ,i think all are fine and good "
l=word_tokenize(text)
print(list(bigrams(l)))

print(list(trigrams(l)))
```
