```python
import spacy
nlp=spacy.load("en_core_web_sm")
m="""Well, I don’t really know why it happened, but it was there, in the middle of the road, and I just stood there for   a moment. It wasn’t bright, nor was it particularly large, yet it seemed to be something that everyone could see, and still, no one said a word. So, as I walked toward it, I thought about how things are often strange, and how we sometimes ignore what is right in front of us. After all, if it is not broken, then why fix it? But then again, if it is broken, shouldn’t we at least try?"""
k=nlp(m)
token=[n for n in k if not n.is_stop and  not n .is_punct]
nouns=[]
verbs=[]
for i in token:
  if i.pos_=='NOUN':
    nouns.append(i)
  if i.pos_=='VERB':
    verbs.append(i)
print("List of NOUNS:",nouns)
print()
print("List of VERBS",verbs)

```
