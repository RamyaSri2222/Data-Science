```python
from numpy import loadtxt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import plot_model
df=loadtxt("pima-indians-diabetes.csv",delimiter=",")
x=df[:,0:8]
y=df[:,8]
model=Sequential()

m=Dense(12,input_shape=(8,),activation="relu")
m1=Dense(8,activation="relu")
m2=Dense(1,activation="sigmoid")
model.add(m)
model.add(m1)
model.add(m2)
model.compile(loss="binary_crossentropy",optimizer="adam",metrics=["accuracy"])
model.fit(x,y,epochs=150,verbose=10)
_,accuracy=model.evaluate(x,y)
print("Accaury:%.2f"%(accuracy*100))
prediction=(model.predict(x)>0.5).astype(int)
for i in range(5):
   print(x[i],y[i])
model.summary()
plot_model(model,to_file="model.png",show_shapes=True,show_dtype=True,show_layer_names=True,expand_nested=True,show_layer_activations=True)
```
