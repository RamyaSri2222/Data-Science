```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, r2_score

# Load data
df = pd.read_csv("homeprices1.csv")
x = df.iloc[:, 0:1]  # feature
y = df.iloc[:, -1]   # target

# Split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)

# Model
reg = LinearRegression()
reg.fit(x_train, y_train)

# Predictions
y_pred = reg.predict(x_test)

# Metrics
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print("Mean Squared Error:", mse)
print("R^2 Score:", r2)

# Sort x_test for proper line plotting
sorted_indices = np.argsort(x_test.values.ravel())
x_test_sorted = x_test.values[sorted_indices]
y_pred_sorted = y_pred[sorted_indices]

# Plot
plt.scatter(x_test, y_test, color="blue", marker="*", label="Actual")
plt.plot(x_test_sorted, y_pred_sorted, color="red", label="Predicted")
plt.xlabel("Home Size (sqft)")
plt.ylabel("Home Price ($)")
plt.title("Home Prices Prediction", color="darkgreen")
plt.legend()
plt.show()
```
