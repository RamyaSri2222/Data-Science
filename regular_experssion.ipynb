```python
import re
text="""Wow! Can you believe it? The rain poured down—hard and fast—while people shouted, laughed, and ran for cover. “What a day,” she said, smiling, even though her shoes were soaked"""
l = re.sub(r"\w+", "", text)
print(l)
```
