```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
df={
    "Company":["Google","Twitter","Google","LinkedIn"],
    "Role":["Data Scientist","Data Analyst","HR","HR"]
    }
df1=pd.DataFrame(df)
print(df1)
model=LabelEncoder()
df1["company1"]=model.fit_transform(df["Company"])
df1["Role1"]=model.fit_transform(df["Role"])

print()
print(df1)
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
df={
    "x":[1,2,3,4,5],
    "y":[3,4,1,2,3]
}
df1=pd.DataFrame(df)
print(df1)
model= MinMaxScaler()
df1["x"]=model.fit_transform(df1[["x"]])
df1["y"]=model.fit_transform(df1[["y"]])
print()
print(df1)
```

```python
from sklearn.preprocessing import FunctionTransformer
d=[[10,20],[30,40],[50,60]]
model=FunctionTransformer()
d1=model.transform(d)
print(d1)
```

```python
from sklearn.preprocessing import FunctionTransformer
import numpy as np
a=[[10,20],[30,40],[50,60]]
f=np.array(a)

def add_ten(f):
  return f+10

model=FunctionTransformer(add_ten)
r=model.transform(f)
print(r)
```
