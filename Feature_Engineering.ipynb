**Numerical**

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
df={
    "Company":["Google","Twitter","Google","LinkedIn"],
    "Role":["Data Scientist","Data Analyst","HR","HR"]
    }
df1=pd.DataFrame(df)
print(df1)
model=LabelEncoder()
df1["company1"]=model.fit_transform(df["Company"])
df1["Role1"]=model.fit_transform(df["Role"])

print()
print(df1)
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
df={
    "x":[1,2,3,4,5],
    "y":[3,4,1,2,3]
}
df1=pd.DataFrame(df)
print(df1)
model= MinMaxScaler()
df1["x"]=model.fit_transform(df1[["x"]])
df1["y"]=model.fit_transform(df1[["y"]])
print()
print(df1)
```

```python
from sklearn.preprocessing import FunctionTransformer
import numpy as np
a=[[10,20],[30,40],[50,60],[70,80]]
f=np.array(a)

def add_ten(x):
  return x+10
ob=FunctionTransformer(add_ten)
model=ob.transform(f)
print(model)
```

```python
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

df=[
    [85,"M","verygood"],
    [95,"F","excellent"],
    [60,"None","good"],
    [np.nan,"M","avrage"],
    [70,"M","good"],
    [np.nan,None,"verygood"],
    [60,"F","verygood"],
    [98,"M","execellent"]
]
cols=["marks","gender","result"]
df2=pd.DataFrame(df,columns=cols)
print(df2)

model=SimpleImputer(missing_values=np.nan,strategy="constant",fill_value=80)

result = df2["marks"].values.reshape(-1,1)
df2["marks"] = model.fit_transform(result)
print()
print(df2)
```

**Categorical**

```python
from sklearn.preprocessing import OrdinalEncoder
from numpy import asarray
df=asarray([["blue"],["green"],["red"]])
encoder=OrdinalEncoder()
model=encoder.fit_transform(df)
print(model)
```

```python
from sklearn.preprocessing import OneHotEncoder
from numpy import asarray
df=asarray([["blue"],["green"],["red"]])
encoder=OneHotEncoder(sparse_output=False)
model=encoder.fit_transform(df)
print(model)
```

```python
from sklearn.preprocessing import OneHotEncoder
from numpy import asarray
df=asarray([["apple"],["banana"],["peer"],["apple"]])
encoder=OneHotEncoder(sparse_output=False)
model=encoder.fit_transform(df)
print(model)
```

```python
from sklearn.preprocessing import OneHotEncoder
from numpy import asarray
df=asarray([["apple"],["banana"],["peer"],["apple"]])
encoder=OneHotEncoder(drop="first",sparse_output=False)
model=encoder.fit_transform(df)
print(model)
```

```python
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

df=[
    [85,"M","verygood"],
    [95,"F","excellent"],
    [60,"M","good"],
    [80,np.nan,"avrage"],
    [70,"M","good"],
    [70,np.nan,"verygood"],
    [60,"F","verygood"],
    [98,"M","execellent"]
]
cols=["marks","gender","result"]
df2=pd.DataFrame(df,columns=cols)
print(df2)

model=SimpleImputer(missing_values=np.nan,strategy="most_frequent",fill_value="80")

result = df2["gender"].values.reshape(-1,1)
df2["gender"] = model.fit_transform(result).ravel()
print()
print(df2)
```
