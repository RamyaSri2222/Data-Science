```python
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB

# Load dataset
df = sns.load_dataset("titanic")

# Features and target
x = df[['sex', 'age', 'pclass', 'fare']]
y = df['survived']

# Convert categorical column to numeric
d = pd.get_dummies(x['sex'], dtype=int)
x = pd.concat([x, d], axis='columns')
x.drop('sex', axis='columns', inplace=True)

# Fill missing values
x['age'] = x['age'].fillna(x['age'].mean())

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=0
)

# Model
model = GaussianNB()
model.fit(x_train, y_train)

# Prediction
print(model.predict(x_test))
```

```python
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB

# Load dataset
df = sns.load_dataset("titanic")

# Select required columns (note: all lowercase)
x = df[['sex', 'age', 'pclass', 'fare']]
y = df['survived']

# Convert categorical column to numeric
sex_dummies = pd.get_dummies(x['sex'], dtype=int)
x = pd.concat([x, sex_dummies], axis=1)
x.drop('sex', axis=1, inplace=True)

# Handle missing values
x['age'] = x['age'].fillna(x['age'].mean())

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=0
)

# Train model
model = GaussianNB()
model.fit(x_train, y_train)

# Predict
print(model.predict(x_test))
```

```python
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
import pandas as pd

df = sns.load_dataset("titanic")

x = df.drop("survived", axis="columns")
y = df["survived"]

d = pd.get_dummies(x["sex"], dtype=int)
x = pd.concat([x, d], axis="columns")
x.drop("sex", axis="columns", inplace=True)

x.age = x.age.fillna(x.age.mean())

# DROP NON-NUMERIC COLUMNS (this fixes the error)
x = x.select_dtypes(include=["number"])

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=0)

mol = GaussianNB()
mol.fit(x_train, y_train)

print(mol.predict(x_test))
```
