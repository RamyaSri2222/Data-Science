```python
from numpy import expand_dims
from tensorflow.keras.utils import load_img
from tensorflow.keras.utils import save_img
from tensorflow.keras.utils import  img_to_array
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
x=load_img("peak.jpeg")
data=img_to_array(x)
sample=expand_dims(data,0)
datagen=ImageDataGenerator()
it=datagen.flow(sample,batch_size=1)

for i in range(9):
  plt.subplot(3,3,1+i)
  batch=next(it)
  image=batch[0].astype("uint8")
  plt.imshow(image)
plt.show()

```

```python
import cv2
import matplotlib.pyplot as plt
image=cv2.imread("artro.webp")
edges=cv2.Canny(image,50,100,3,L2gradient= True)
plt.figure()
plt.title("Spider")
plt.imsave("dancing-spider-canny",edges,cmap="gray",format='png')
plt.imshow(image)
plt.imshow(edges,cmap="gray")
plt.show
plt.show()
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image=cv2.imread("as.jpeg")

gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)

edged=cv2.Canny(gray,30,200)


contours,hierarchy=cv2.findContours(edged,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_NONE)

cv2_imshow(edged)
cv2.drawContours(image,contours,-1,(0,255,0),3)

cv2_imshow(image)
```
