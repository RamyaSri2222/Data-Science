```python
from collections import Counter
import spacy
nlp=spacy.load("en_core_web_sm")
text="""Well, you know, it was kind of a strange afternoon, and I was just sitting there, thinking about all the things that I should have done but didn’t. The clock kept ticking, and it was, in a way, both comforting and a little bit annoying. I mean, it’s not like I had somewhere important to be, but still, the sound was there, over and over again.

And then, for no particular reason, I decided to make some tea; although, honestly, I wasn’t even that thirsty. So, I filled the kettle, waited for it to boil, and stared out of the window while the water heated up. There were people walking by, and they seemed to be in a hurry, as if they had plans, or at least somewhere to go.

By the time I finally sat back down, with the cup in my hands, I had almost forgotten why I got up in the first place. It wasn’t about the tea, or the clock, or even the quiet room—it was just one of those moments where nothing special happens, and yet, somehow, everything feels oddly complete."""
l=nlp(text)
tok=[token for token in l if not token.is_stop and not token.is_punct]
model=Counter(tok)

print(model.most_common(6))
```
