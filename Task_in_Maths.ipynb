# **Asymmetry-Left**

```python
import matplotlib.pyplot as plt
data=[10,20,30,40,50,60,10,20,30]
plt.boxplot(data)
plt.title("Asymmetry-left")
plt.show()
```

# **Asymmetry-Right**

```python
import matplotlib.pyplot as plt
data=[50,30,20,40,50,60,10]
plt.boxplot(data)
plt.title("Aymmetry-Right")
plt.show()
```

# **Symmetric**

```python
import matplotlib.pyplot as plt
data=[5,22,38,50,10,20,30,40,100,10,20]
b=[5,10,20,30,40,50,60]
plt.hist(data,bins=b)
plt.title("symmetric histogram")
plt.show()
```

```python
import matplotlib.pyplot as plt

data_left = [2,3,4,5,6,7,8,9,9,9,10,10,10]
plt.hist(data_left, bins=5)
plt.title("Left Asymmetric Histogram")
plt.show()
```
